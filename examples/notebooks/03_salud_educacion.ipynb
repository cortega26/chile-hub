# Salud y educacion por comuna

Cuenta establecimientos de salud y educacionales por comuna usando el catalogo territorial.

```python
# Instala chile-hub si el entorno no lo trae (p. ej. Google Colab)
try:
    import chile_hub  # noqa: F401
except ImportError:
    %pip install -q chile-hub
```

```python
from chile_hub import ChileHub

hub = ChileHub()
comunas = hub.load_polars("comunas")
salud = hub.load_polars("establecimientos_salud")
educacion = hub.load_polars("establecimientos_educacionales")
```

```python
salud_por_comuna = salud.group_by("codigo_comuna").len("establecimientos_salud")
educacion_por_comuna = educacion.group_by("codigo_comuna").len("establecimientos_educacionales")

resumen = (
    comunas.join(salud_por_comuna, on="codigo_comuna", how="left")
    .join(educacion_por_comuna, on="codigo_comuna", how="left")
    .fill_null(0)
    .select(
        "codigo_comuna",
        "nombre_comuna",
        "nombre_region",
        "establecimientos_salud",
        "establecimientos_educacionales",
    )
)
resumen.sort("establecimientos_salud", descending=True).head(15)
```
